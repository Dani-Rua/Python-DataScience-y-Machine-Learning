## Resolución

Crea una agenda vacía

```python
agenda = {}
```

En la siguiente celda, incluye el código para ejecutar las siguientes instrucciones:
- Mostrar en pantalla las opciones
- Pedirle al usuario que elija las opciones
- Crear una estructura de control (*if*, *elif* y *else*) que ejecute diferentes bloques de código según las opciones elegidas.

```python
print("Cuál de estas funciones prefieres:\n 1.Añadir un contacto.\n 2.Buscar un contacto.\n 3.Editar un contacto.\n 4.Eliminar un contacto.\n 5.Motrar todos los contactos.")
opciones = int(input())
```

```python
if opciones == 1:
    print("Ingrese el nombre a añadir: ")
    nombre = input()
    print("Ingrese su número de celular: ")
    num_celular = input()
    agenda[nombre] = num_celular
    print(agenda)
```

```python
if opciones == 2:
    buscar_nombre = input("Ingrese el nombre a buscar: ")
    if buscar_nombre in agenda:
        print(f"El nombre que busca es {buscar_nombre} y su número de teléfono es {agenda[buscar_nombre]} si esta en la agenda")
    else:
        print("Ese nombre no se encuentra en la agenda")
```

```python
if opciones == 3:
    nombre = input("Ingrese el contacto a modificar: ")
    num_modif = input("Ingrese el nuevo número del contacto : ")
    agenda[nombre] = num_modif
    print(f"El usario modificado es {nombre} y su nuevo número es {num_modif}")
    print(f"Agenda actualizada: {agenda}")
```

```python
if opciones == 4:
    nombre = input("Ingrese el nombre del usuario a eliminar: ")
    del agenda[nombre]
    print(f"El usario eliminado es {nombre}")
    print(agenda)
    
```

```python
if opciones == 5:
    print("Lista de contactos: ")
    for nombre, num_celular in agenda.items():
        print(f"Nombre: {nombre} | Teléfono: {num_celular}")
```
