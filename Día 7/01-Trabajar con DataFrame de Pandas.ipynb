```python
import pandas as pd
```

```python
datos = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Sara'],
    'Edad': [25, 30, 22, 27],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Bilbao']
}

df = pd.DataFrame(datos)
```

```python
df
```

```python
# Agregar una nueva columna

df['Salarios'] = [30000, 45000, 38000, 32000]
df
```

```python
# Modificar la columna de un df
df['Salarios'] = df['Salarios'] + 2000
df
```

```python
# Seleccionar una columna y almacenarla en una variable
nombres = df['Nombre']
nombres
```

```python
# Aplicar un método de filtrado
mayores_25 = df[df['Edad'] > 25]
mayores_25
```

```python
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Sara'],
    'Edad': [25, 30, 22, 27],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Bilbao']
}

df = pd.DataFrame(data)
edades = df['Edad']
df_filtrado = df[edades >= 25]
df_filtrado
```
