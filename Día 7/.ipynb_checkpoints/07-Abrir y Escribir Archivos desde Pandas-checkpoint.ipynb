```python
import pandas as pd
```

```python
# Se muestra un rango de fecha por día
fechas = pd.Series(pd.date_range('20240101', periods=6))
fechas
```

```python
type(fechas[0])
```

```python
# Este es el parámetro que viene por defecto para freq por eso no salta años ni meses, solo por día a día
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='D'))
fechas
```

```python
# Va a hacer saltos mes por mes
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='ME'))
fechas
```

```python
# Va a hacer saltos año en año
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='YE'))
fechas
```

```python
# Va a hacer saltos hora en hora
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='h'))
fechas
```

```python
# Va a hacer saltos minuto en minuto
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='min'))
fechas
```

```python
# Va a hacer saltos de a días
fechas = pd.Series(pd.date_range('20240101', periods=6, freq='5D'))
fechas
```
