```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Col1': [100, 200, 300],
    'Col2': [400, 500, 600],
    'Col3': [700, 800, 900]
}, index=['fila1', 'fila2', 'fila3'])
df
```

```python
# Loc nos permite filtrar por filas o columnas basandose en sus etiquetas
# Nos sirve para seleccionar filas y sus respectivas columnas
df.loc['fila1']
```

```python
type(df.loc['fila1'])
```

```python
# Para traer dos filas se debe ingresar doble llaves cuadradas
df.loc[['fila1', 'fila3']]
```

```python
type(df.loc[['fila1', 'fila3']])
```

```python
# Buscar las filas como segmentación
df.loc['fila2':'fila3']
```

```python
# Pasar una lista de booleanos
df.loc[[False, True, False]]
```

```python
# Algunas filas que cumplan una condición particular
df.loc[df['Col1'] > 150]
```

```python
# Y si queremos trar columnas en ves de filas
df.loc[:, ['Col1', 'Col2']]
```

```python
# iloc se basa en las pocisiones numéricas
df.iloc[0]
```

```python
# Los : puntos indican las filas, ahí le decimos que me pase todas las filas y despues de la , cuantas columnas quiero que me traiga
df.iloc[:, [0,1]]
```

```python
df.iloc[1:3]
```

```python
df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carmen'],
    'Edad': [25, 30, 22],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia']
}, index=['fila1', 'fila2', 'fila3'])

df.loc['fila2']
```

```python
import pandas as pd
df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [5, 6, 7, 8],
    'C': [9, 10, 11, 12],
    'D': [13, 14, 15, 16]
})

resultado = df.iloc[:2, [2,3]]
resultado
```
