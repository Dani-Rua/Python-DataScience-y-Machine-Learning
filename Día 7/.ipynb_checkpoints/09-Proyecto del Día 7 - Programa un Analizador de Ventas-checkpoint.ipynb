```python
# 2.Fusión de Datos:
ventas_concatenada = pd.concat([ventas1, ventas2])
ventas_concatenada
```

```python
# 3.Tratamiento de Datos:
ventas_concatenada.shape
```

```python
ventas_concatenada.isnull().sum()
```

```python
ventas_concatenada.info
```

```python
ventas_concatenada.describe()
```

```python
# 4.Análisis de Ventas:
# ¿Cuál es el producto más vendido?
mas_vendido = ventas_concatenada['Producto'].value_counts()
mas_vendido
```

```python
# ¿Cuál es el mes con mas ventas?
mes_mas_ventas = ventas_concatenada.groupby('Fecha')['Total Venta'].sum().sort_values(ascending=False)
mes_mas_ventas
```

```python
# 5.Datos Agrupados:
agrupar_categoria_producto = ventas_concatenada.groupby('Producto')['Total Venta'].sum().sort_values(ascending=False)
agrupar_categoria_producto
```
