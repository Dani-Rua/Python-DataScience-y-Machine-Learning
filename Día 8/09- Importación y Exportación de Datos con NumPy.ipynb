```python
# Esto se hace para delimitarlos por , ya que es un archivo csv y si no se hace da un error por en la linea anterior
array = np.genfromtxt(ruta, delimiter=',')
array
```

```python
# Ya vismo anteriormente que los espacios son nan, pero no los queremos de esa forma, entonces los llevamos a cero
array = np.genfromtxt(ruta, delimiter=',', filling_values=0)
array
```

```python
# Vamos a eliminar ignorando los encabezados de la columna
array = np.genfromtxt(ruta, delimiter=',', filling_values=0, skip_header=1)
array
```

```python
# Cambiar a integer
array = np.genfromtxt(ruta, delimiter=',', filling_values=0, skip_header=1, dtype=int)
array
```
