```python
import numpy as np
```

```python
array1 = np.array([1,2,3])
array1
```

```python
array2 = np.array([[0], [10], [20], [30]])
array2
```

```python
# Broadcasting, extender operaciones. Vamos a sumar los arrays que tienen diferente forma
broadcast_suma = array1 + array2
broadcast_suma
```

```python
broadcast_multi = array1 * array2
broadcast_multi
```

```python
# Funciones universales en NumPy, se aplica vectorizada
a = np.array([1,2,3])
b = np.array([4,5,6])
```

```python
# Nos devuelve un solo array de la suma
resultado = np.add(a, b)
resultado
```

```python
# Nos devuelve un solo array de la resta
resultado = np.subtract(a, b
resultado
```

```python
# Nos devuelve un solo array de la multiplicación
resultado = np.multiply(a, b)
resultado
```

```python
# Nos devuelve un solo array de la división
resultado = np.divide(a, b)
resultado
```

```python
# Funciones matemáticas universales, conocer el exponencial
resultado2 = np.exp(a)
resultado2
```

```python
# Funciones matemáticas universales, conocer el logaritmo natural
resultado3 = np.log(a)
resultado3
```

```python
resultado4 = np.log10(a)
resultado4
```

```python
# Funciones matemáticas universales, conocer la raíz cuadrada
resultado5 = np.sqrt(a)
resultado5
```
