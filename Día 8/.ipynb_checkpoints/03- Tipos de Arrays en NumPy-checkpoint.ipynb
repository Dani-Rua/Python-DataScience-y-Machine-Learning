```python
import numpy as np
```

```python
# Es un array unidimensional, que tiene solo una dimensión, es el largo por eso se llama de esa forma
array = np.array([1,2,3])
array
```

```python
len(array)
```

```python
# Cual es la forma, notamos que tiene el largo, pero el alto, por eso despues de la , esta vacío
array.shape
```

```python
# Array de dos dimensiones
array_2d = np.array([[1,2,3], [4,5,6]])
array_2d
```

```python
# El 2 indica que tiene dos elementos, que indica que es cada uno de los conjustos [1, 2, 3] y [4, 5, 6] por esos son dos. Este es el largo
len(array_2d)
```

```python
# Consultar su forma. 2 es el alto o las columnas y 3 es el largo o las filas
array_2d.shape
```
