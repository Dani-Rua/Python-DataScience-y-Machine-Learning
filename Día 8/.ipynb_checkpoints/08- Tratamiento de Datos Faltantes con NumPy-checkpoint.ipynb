```python
import numpy as np
```

```python
# Crear un array con un dato faltente, esto nos permite reservar un espacio
array = np.array([1, 2, np.nan, 4, 5])
array
```

```python
# Cómo ver que hay un nan
np.isnan(array)
```

```python
# Cómo manejar nuestros elementos no válidos, el promedio no se puede sacar porque tiene valores nan
np.mean(array)
```

```python
# Funciones que ignoran números no válidos
np.nanmean(array)
```

```python
# Sustituir los valores no válidos, se usa mucho
array_con_0 = np.where(np.isnan(array), 0, array)
array_con_0
```

```python
# Reemplazar con un primedio
promedio = np.nanmean(array)
array_con_promedio = np.where(np.isnan(array), promedio, array)
array_con_promedio
```

```python
# Cómo eliminar datos faltantes. La tílde invertida ~ hace que muestre el array completo sin el nan, porque si lo dejamos sin eso muestra
# el nan y eso no es lo que queremos eliminar, porque si muestra el nan quiere decir que eliminamos los números.
array_filtrado = array[~np.isnan(array)]
array_filtrado
```

```python
array = np.array([3, np.nan, 7, np.nan, 0, 4.5, np.nan])

sustituir_nans = np.where(np.isnan(array), 0, array)
sustituir_nans
```
