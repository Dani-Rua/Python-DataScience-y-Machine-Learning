```python
import numpy as np
```

```python
x = [1,2,3]
y = [1,2,3]
z = [1,2,3]
```

```python
# Vamos a concatenarlos, quedan en una sola dimensión
np.concatenate([x, y, z])
```

```python
# Vamos a probar en que eje lo queremos tener sería con index, que se concatenen en el eje 1
a = [[1, 2],
    [3, 4]]
b = [[5, 6],
    [7, 8]]
```

```python
np.concatenate([a, b], axis=1)
```

```python
# Vamos a probar en que eje lo queremos tener sería con index, que se concatenen en el eje 0
np.concatenate([a, b], axis=0)
```

```python
# Cambiamos las forma del array
array_concatenado = np.concatenate([a, b], axis=0)
array_concatenado
```

```python
# Conocer la forma, 4 filas por 2 columnas
array_concatenado.shape
```

```python
# Podemos hacer que se reorganice, de 2 filas por 4 columnas
array_reformado = array_concatenado.reshape(2, 4)
array_reformado
```

```python
# Sumar arrays
array_sumados = array_reformado + array_reformado
array_sumados
```

```python
# Multiplicados arrays
array_multi = array_reformado * array_reformado
array_multi
```

```python
# Raiz cuadrada en arrays
array_raices = np.sqrt(array_multi)
array_raices
```
