```python
# Comparar la velocidad de array de NumPy vs una lista, lo hacemos con 1 millon de elementos
# Acá creo una lista con un millon de elementos
lista_grande = list(range(1000000))

# Acá creo un array de NumPy con un millon de elementos
array_grande = np.array(lista_grande)
```

```python
type(lista_grande)
```

```python
type(array_grande)
```

```python
# Medimos la ejecución del array y lista. Vamos a hacer el cuadrado de un millon de número esos nos permite medir el tiempo
inicio_lista = time.time()
for i in lista_grande:
    i ** 2
fin_lista = time.time()
print("Fin tiempo lista: ", fin_lista - inicio_lista)
```

```python
# Esto lo permite porque array tiene funciones vectorizadas
inicio_array = time.time()
array_grande ** 2
fin_array = time.time()
print("Tiempo de array: ", fin_array - inicio_array)
```
