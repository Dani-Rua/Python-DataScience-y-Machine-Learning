```python
import pandas as pd

data = {
        "Id_producto": [1001, 1002, 1003, 1003],
           "Cantidad_vendida": [30, None, 25, 25],
            "Precio": [20.5, 15.0, None, 22.5]
       }

df = pd.DataFrame(data)
df
```

```python
#Se realiza para ver los primeros datos de la tabla
df.head()
```

```python
# Se realiza para conocer la información del dataframe
df.info()
```

```python
# Identifica los nulos
df.isnull()
```

```python
# Suma los nulos que hay en el dataframe
df.isnull().sum()
```

```python
# Elimina los nulos por completo con su fila
df_eliminados = df.dropna()
df_eliminados
```

```python
# Saca un promedio de la serie mencionada para llenar el campo null
valores_nuevos = {"Cantidad_vendida": 0, "Precio": df["Precio"].mean()}
df_rellenados = df.fillna(valores_nuevos)
df_rellenados
```

```python
# Se realiza un cambio de tipo para ir limpiando los datos
df_rellenados["Cantidad_vendida"] = df_rellenados["Cantidad_vendida"].astype(int)
df_rellenados
```

```python
# Se elimina los datos duplicados de una serie en específico
df_unicos = df_rellenados.drop_duplicates(subset=["Id_producto", "Cantidad_vendida"])
df_unicos
```
