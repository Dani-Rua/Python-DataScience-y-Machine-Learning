```python
import numpy as np
import matplotlib.pyplot as plt

# Esto permite no poner al final el shown sino con solo poner un ; muestra el gráfico
%matplotlib inline
```

```python
# Cómo crear un Scatter plot, linspace conjunto de datos en un rango
x = np.linspace(0, 10, 30)
# Sin se usa para hacer el seno de x en este caso, lo hicimos para inventarnos 30 números al azar
y = np.sin(x)
```

```python
# Acá logramos un gráfico de líneas
plt.plot(x, y);
```

```python
# Para generar un gráfico scatter que es con puntos
plt.plot(x, y, 'o');
```

```python
aleatorio = np.random.RandomState(0)
marcadores = ['o', 'x', '+', 'v', 's', 'd']
```

```python
for marcador in marcadores:
    plt.plot(aleatorio.rand(5), aleatorio.rand(5), marcador)
```

```python
plt.plot(x, y, ':ob');
```

```python
# Los puntos son pentágonos
plt.plot(x, y, '-p');
```

```python
# Los puntos son pentágonos y color rojo
plt.plot(x, y, '-p', color='red');
```

```python
# Los puntos son pentágonos y color rojo, el tamaño de los pentágonos
plt.plot(x, y, '-p', color='red', markersize=15);
```

```python
# Los puntos son pentágonos y color rojo, el tamaño de los pentágonos, linea mas gruesa
plt.plot(x, y, '-p', color='red', markersize=15, linewidth=4);
```

```python
# Los puntos son pentágonos y color rojo, el tamaño de los pentágonos, linea mas gruesa
plt.plot(x, y, '-p', color='red', markersize=15, linewidth=4, markerfacecolor='white');
```

```python
# Creamos gráfico Scatter con el método scatter
plt.scatter(x, y)
```

```python
# Mas propiedades de los Scatter, c= colores de los puntos, s= al tamaño de los puntos, alpha = la transparencia,
# cmap = mapa de colores
aleatorio = np.random.RandomState(0)
plt.scatter(aleatorio.randn(100), 
            aleatorio.randn(100),
           c=aleatorio.randn(100),
           s=aleatorio.randn(100)*1000,
           alpha=0.3,
           cmap='viridis')
# colorbar crea una barra de referencias
plt.colorbar();
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.scatter(x, y, color='red', s=50)
```
