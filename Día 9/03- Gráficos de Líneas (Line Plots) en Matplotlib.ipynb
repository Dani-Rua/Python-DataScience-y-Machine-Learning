```python
import pandas as pd
import matplotlib.pyplot as plt

# Esto permite no poner al final el shown sino con solo poner un ; muestra el gráfico
%matplotlib inline
```

```python
# Siempre creamos una figura y un axes antes de crear una figura
fig = plt.figure()
axs = plt.axes()
```

```python
# Lineas de referencia
plt.grid()
```

```python
# Creamos una lista
x = [1,2,3,5,7,8,9]
y = [1,6,9,3,5,7,9]
```

```python
# Plot es una función que puedes usar para crear gráficos y tienes mas funciones
plt.plot(x, y, "ro-")
plt.grid()
```

```python
# Ponerle color y forma a las lineas y ejes
plt.plot(x, y, "b*:")
# configurar el valor máximo o mínimo, personalizar los ejes
plt.axis([0, 10, 0, 10])
plt.grid()
```
